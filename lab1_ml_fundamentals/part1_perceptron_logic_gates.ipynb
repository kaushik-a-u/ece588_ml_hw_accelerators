```python
#Task3
def XOR(a, b):
    or_out = perceptron(a, b, 1, 1, -0.5)
    nand_out = perceptron(a, b, -1, -1, 1.5)
    return perceptron(or_out, nand_out, 1, 1, -1.5)

print("XOR Gate:")
for a in [0,1]:
    for b in [0,1]:
        print(a, b, XOR(a,b))
```

```python
#Task 4
def NOT(x):
    return perceptron(x, 0, -1, 0, 0.5)

def MUX(a, b, s):
    not_s = NOT(s)
    term1 = perceptron(a, not_s, 1, 1, -1.5)
    term2 = perceptron(b, s, 1, 1, -1.5)
    return perceptron(term1, term2, 1, 1, -0.5)

print("2-to-1 MUX Truth Table:")
for a in [0,1]:
    for b in [0,1]:
        for s in [0,1]:
            print(f"A={a}, B={b}, S={s} => Y={MUX(a,b,s)}")
```

```python
#Task 5
#Part1
def sigmoid(x):
    return 1/(1+np.exp(-x))

W1 = np.array([[1,1],[1,1]])
b1 = np.array([-0.5, 1.5])
W2 = np.array([[1],[1]])
b2 = np.array([-1.5])

def forward(x):
    h = sigmoid(np.dot(x,W1.T)+b1)
    out = sigmoid(np.dot(h,W2)+b2)
    return out

X = np.array([[0,0],[0,1],[1,0],[1,1]])
for i in range(4):
    print(X[i], forward(X[i]))
```

```python
# TODO: Train XOR with PyTorch
import torch
import torch.nn as nn
import torch.optim as optim

X_t = torch.tensor([[0,0],[0,1],[1,0],[1,1]], dtype=torch.float32)
y_t = torch.tensor([[0],[1],[1],[0]], dtype=torch.float32)

model = nn.Sequential(
    nn.Linear(2,4),
    nn.Tanh(),
    nn.Linear(4,1),
    nn.Sigmoid()
)

criterion = nn.BCELoss()
optimizer = optim.Adam(model.parameters(), lr=0.1)

# TODO: Run training loop and print loss every 500 epochs
for epoch in range(2000):
    optimizer.zero_grad()
    outputs = model(X_t)
    loss = criterion(outputs,y_t)
    loss.backward()
    optimizer.step()
    if epoch % 500 == 0:
        print(f"Epoch {epoch}, Loss: {loss.item():.4f}")

print("Final Predictions:", model(X_t).detach().numpy())
```
